```python
from scitacean import Dataset, Profile
from scitacean.testing.client import FakeClient
from scitacean.testing.transfer import FakeFileTransfer

from scicat_widget import DatasetUploadWidget


def source_folder(proposalIds, instrumentIds):
    if not proposalIds or not instrumentIds:
        return None
    return f"/data/{'-'.join(proposalIds)}/{'-'.join(instrumentIds)}"


def owner_group(proposalIds):
    return "-".join(proposalIds) or None


def creation_location(proposalIds):
    return f"loc-{'-'.join(proposalIds)}" if proposalIds else None


transfer = FakeFileTransfer()
profile = Profile(
    url="https://fake.scicat",
    file_transfer=transfer,
    field_factories={
        "sourceFolder": source_folder,
        "ownerGroup": owner_group,
        "creationLocation": creation_location,
    },
)
client = FakeClient.without_login(profile)

initial = Dataset(type="raw", proposal_ids=["p.123"], instrument_ids=["instr.9"])
DatasetUploadWidget(client, initial=initial, non_editable=["sourceFolder"])
```

```python
initial = Dataset(
    type="raw",
    proposal_ids=["p.123"],
    instrument_ids=["instr.9"],
    source_folder="/initial",
)
DatasetUploadWidget(client, initial=initial, locked=["sourceFolder", "ownerGroup"])
```
